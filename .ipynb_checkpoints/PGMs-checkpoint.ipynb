## **Intuición**

Los PGMs se usan en inteligencia artificial para el diagnostico de enfermedades y la incidencia que podria tener sus respectivos tratamientos, para la segmentación de imagenes en donde s epueda reconocer que segmento perteneces a que parte de la imagen.

Lo que tienen en común estos ejemplos es que por lo general tiene bastante variables a conciderar, bastante incertidumbre sobre la respuesta correcta.

## **Definición**

Los PGMs son estructuras aptas para gestionar aplicaciones

Un ejemplo muy usual de PGM es la red Bayesiana (como ejemplo de este tipo de red esta la CPC
usada para el diagnóstico medico) la cual se puede representar mediante un grafo dirigido, otro 
también es la red de Markov (un ejemplo es lo de segmentación de imágenes) la cual por el  contrario 
utiliza grafos no dirigidos.

Factor
¿Que es? Es una función o una tabla, que manda un conjunto de variables aleatorias (sus realizaciones) a un valor de los reales, El conjunto de vairables aleatorias se suele llamar el ambito o alcance del factor ,en ese caso, las distribuciones de probabilidad conjuntas son un ejemplo abstante facil de factores, una medida sin normalizar también es un factor, un facotr que se tuilizará aca es la llamada distribución de probabilidad condicional.

Factores Generales: Pueden ser tomados de cualquier manera y no necesariamnte deben conducir a probabilidades.

Operaciones entre factores:

Producto: Es el producto normal de funciones solo que el nuevo ambito es l aunicon de los ambitos que son multiplicados.
Marginalizacion de factores: Es identifca a la marginalizacion de probabilidad
Reduccion de factores: Es identifca a al reduccion probabilista también pero sin la nornalización.

¿Por que usamos factores? La forma en la que se define una distribución de probabilidad exponencialmente grande es trabajar con muchos pequeños trozos e integrarlos después mediante multiplicación de factores
